# Time Series Anomaly Detection - Exploration

This notebook explores the Yahoo Webscope S5 A1Benchmark dataset.
We visualize the time-series signal and labeled anomalies to
understand the structure of the data.

```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use("default")
```

```python
df = pd.read_csv("C:/Users/owen/projects/time-series-anomaly-detection/data/raw/A1Benchmark/real_1.csv")
df.head()
```

```python
df.info()
```

## Dataset Columns

- **timestamp**: Time index of the observation
- **value**: Observed value of the time series
- **is_anomaly**: Binary label indicating whether the point is anomalous

```python
plt.figure(figsize=(12,4))
plt.plot(df["value"])
plt.title("Raw time-series signal")
plt.xlabel("Time")
plt.ylabel("Value")
plt.show()
```

```python
anomalies= df[df["is_anomaly"] == 1]

##Visualizing the anomalies
plt.figure(figsize=(12,4))
plt.plot(df["value"], label="signal")
plt.scatter(anomalies.index, anomalies["value"],
            color="red", label="anomaly", s=20)
plt.legend()
plt.title("Time-series with labeled Anomalies")
plt.show()
```

## Observations

- The time series shows non-stationary behavior.
- Anomalies appear as sharp deviations from the local pattern.
- The dataset provides ground-truth labels, which gives supervised evaluation.
